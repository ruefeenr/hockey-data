```python
## import csv
csv_full = pd.read_csv("rawData/canada_v_hcd.csv")
csv_full
```

```python
# Filter out only scenarios where powerplay
data = csv_full[(csv_full["TeamStrength"] == "5v4") | (csv_full["TeamStrength"] == "4v5")]
data
```

Can jump down to autoparse

```python
blueline_rows = data[data["EventType"] == "BluelineCrossing"]
blueline_rows = blueline_rows.index
blueline_rows # list which contains the index number of every blueline row

shot_rows = data[data["EventType"] == "Shot"]
shot_rows = shot_rows.index
shot_rows = shot_rows.tolist()

```

```python
bl_start_end = []
for index,value in enumerate(blueline_rows):
    if index == len(blueline_rows)-1:
        break
    bl_start_end.append(
        [value, int(blueline_rows[index+1])]
    )

bl_start_end
blueline_data = pd.DataFrame(bl_start_end, columns=["bl_start_index", "bl_end_index"])
```

```python
#blueline_data["Start_matchclock"] = data[start]
for index, row_data in blueline_data.iterrows():
    st = data["MatchClock"][row_data["bl_start_index"]]
    blueline_data.loc[index,"start_matchClock"] = st

    en = data["MatchClock"][row_data["bl_end_index"]]
    blueline_data.loc[index, "end_matchClock"] = en

blueline_data["playtime_seconds"] = blueline_data['end_matchClock'] - blueline_data["start_matchClock"]

blueline_data = blueline_data[blueline_data["playtime_seconds"]>= 5]

blueline_data.reset_index(inplace= True, drop=True)
blueline_data
```

```python
## shots
# Create intervals from the start and end indices
intervals = pd.IntervalIndex.from_arrays(
    blueline_data["bl_start_index"],
    blueline_data["bl_end_index"],
    closed="neither", # greater than start and less than end
)

# Check if any shot falls inside each interval
shot_mask = pd.Series(
    [any(s in interval for s in shot_rows) for interval in intervals],
    index=blueline_data.index,)

blueline_data = blueline_data[shot_mask]
blueline_data
```

## new direction - New Autoparse
- go to each shot
- go backwards 5 seconds or until you get a faceoff or blueline crossing
    - if uninteruped, caputure that as a successful scenario
- repeat

```python
## just make sure to ingest the entire CSV first.

shot_rows = data[data["EventType"] == "Shot"]
shot_rows = shot_rows.index
shot_rows = shot_rows.tolist()
```

```python
n_seconds = 5
scenario_number = 1
game_name = "Fribourg-HC_Dynamo"

for single_shot in shot_rows:
    clock_at_shot = data.loc[single_shot, "MatchClock"]
    period = data.loc[single_shot, "Period"]
    play_data = data[(data["MatchClock"] >= clock_at_shot - n_seconds) & (data["MatchClock"] <= clock_at_shot) & (data["Period"] == period)] #accesses data which occurs n_seconds prior to shot event.


    if ("BluelineCrossing" in play_data["EventType"].values) | ("Faceoff" in play_data["EventType"].values): #removes scenarios which break structure rules.
        print("found BluelineCrossing or Faceoff. No export")
        continue

    else:
        play_data.to_csv(f"autoparse_csv/{game_name}_shotData{scenario_number}.csv") #exports scenarios which fit structure rules.
        print(f"Exported Scenario {scenario_number} to CSV")
        scenario_number = scenario_number+1
```
